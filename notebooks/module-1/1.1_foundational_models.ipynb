## Initialising and invoking an agent
Interestingly enough one can wrapp up an already-created chat model in an agentic model since the `model` variable is of `ChatBaseModel` type.

Furthermore, direct inputation of the model's name will work with specific string identifers as the following one (`agentic_model:str = "google_genai:gemini-3.8-flash; create_agent(agentic_model)`) 

```python
from langchain.agents import create_agent
agent = create_agent("google_genai:gemini-3.5-flash")
```

```python
# Importing the `agents` langchain module that does this
from langchain.agents import create_agent
from langchain.chat_models import init_chat_model

# Declare variables that are descriptive to the preffered model 
# ... and init the chat model;
model_name:str = 'gemini-3.5-flash'
model_provider:str = 'google_genai'
chat_model = init_chat_model(model=model_name,
                             model_provider=model_provider,
                             configurable_fields='any',
                             temperature=2)

# Now, pluging this chat_model var in create_agent to get an agent out of it;
gemini_agent = create_agent(chat_model)
```

# Handling conversations with an Agent
Conversating with an agent becomes a slightly different job with regards to the output's syntax that changes significantly;

`HumanMessage`:  A `HumanMessage` is a message that is passed in from a user to the model.

`AIMessage`: An `AIMessage` is returned from a chat model as a response to a prompt. Its payload can also carry the history of the chat or even fabricated answers to further indirectly oblige the model behave as expected;

```python
# Import the message related functions...
from langchain_core.messages import HumanMessage, AIMessage
payload1:str = "How do I calculate the maximum drawdown on a leveraged position before a margin call is triggered?"
receivedload1:str =  "My man, chill down and do not waste energy on manual math. Better set up an automated stop-loss to mitigate the risk."
payload2:str = "I tried and the market gapped right past my exit price ;("
response = gemini_agent.invoke({"messages":[HumanMessage(content=payload1),
                                           AIMessage(content=receivedload1),
                                           HumanMessage(content=payload2)]})

pprint(response)
```

```python
print(response['messages'][-1].content)
```

## Streaming Output

A rather more effective dialogue we could have changes the rule of the game, from `direct imputation` to `streaming the prompt` such that the model digest tokens, each corresponding to a specific word in your prompt. The `token` as shown in the loop bellow is a message chunk with token content while `metadata` servers as a dictionary identifer encompasing details such as the node that is producing the token

```python
payload:dict = {"messages":[HumanMessage(content="Tell me about how the first public and official trade unfolded")]}
for token, metadata in gemini_agent.stream(payload, stream_mode="messages"):
    if token.content:
        print(token.content, end="", flush=True)
```
