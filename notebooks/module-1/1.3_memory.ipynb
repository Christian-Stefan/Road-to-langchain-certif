## No memory vs Memory

### No memory

```python
# Import library containing all chat models
from langchain.chat_models import init_chat_model
# Import library for creatign agents
from langchain.agents import create_agent
# Import library containing prompt type functions
from langchain.messages import AIMessage, HumanMessage, SystemMessage
# Declaring questions meant to support the demonstration
question:str = "What is 1+1 equal to?"
follow_up_question:str = "Do you rembmer what I just asked you one thread ago?"

# 1. Create the chat model
det_md:list = ["google_genai","gemini-2.5-flash"]
chat_model_instance = init_chat_model(model_provider=det_md[0],
                                      model=det_md[1])
# 2. Wrap the chat model instance in an agentic-shell
gemini_agent = create_agent(chat_model_instance)


# 3. Ask the previously defined questions and display the agent's answer.
answer1 = gemini_agent.invoke({"messages":question})
answer2 = gemini_agent.invoke({"messages":follow_up_question})
from pprint import pprint
pprint(answer1['messages'])
pprint(answer2['messages'])
```

### Memory

```python
# Import the threading creation module
from langgraph.checkpoint.memory import InMemorySaver
memory_on_agent = create_agent(model=chat_model_instance,
                               checkpointer=InMemorySaver())

# 1. Define the minimum viable required threading protocol as a default dict
threading_protocol:dict = {"configurable":{"thread_id":"1"}}

# 2. Try again
response1 = memory_on_agent.invoke({"messages":question},threading_protocol)
response2 = memory_on_agent.invoke({"messages":follow_up_question}, threading_protocol)

# 3. Display...
pprint(response1["messages"])
pprint(response2["messages"])
```
